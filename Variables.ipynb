#### What is Machine Learning?

Machine Learning is a branch of Artificial Intelligence (AI) that allows computers to learn from data, make decision and improve over time without being directly programmed for every task.

#### Name the three main types of Machine Learning.

Supervised Learning
Unsupervised Learning
Reinforcement Learning

#### What is Supervised Learning?

Supervised learning is a type of machine learning where the computer learns from labeled data. This means the input data already has the correct answers. The model studies these examples and learns how to match inputs with outputs. After training, it can predict answers for new data. 

For example, it can learn to identify whether an email is spam or not by looking at many labeled examples.

#### What is Unsupervised Learning?

Unsupervised learning is a type of machine learning where the computer learns from data without labels. This means there are no correct answers given. The system tries to find patterns, groups or relationships in the data by itself. 

For example, it can group similar customers together based on their buying behavior.

#### What is Reinforcement Learning?

Reinforcement learning is a type of machine learning where an agent learns by interacting with an environment. It learns through trial and error. When it makes a good decision, it gets a reward and when it makes a bad decision, it gets a penalty. Over time, it learns the best actions to get the highest reward. 

For example, it is used in game playing AI. It learns how to win by playing again and again.
