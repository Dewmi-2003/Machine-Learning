```python
X = df[["Age","MonthlySpend"]]
```

```python
wcss = []
l = 1 # Initialize l, assuming it should start from 1
for kin in range(l,11):
    model = KMeans(n_clusters=kin, random_state=42)
    model.fit(X)
    wcss.append(model.inertia_)
```

```python
optimal_k = 3 # Assuming 3 clusters for demonstration, but you would choose this after plotting wcss
model = KMeans(n_clusters=optimal_k, random_state=42)
df['Cluster'] = model.fit_predict(X)

plt.scatter(df["Age"], df["MonthlySpend"],
c=df["Cluster"], cmap="viridis")
plt.xlabel("Age")
plt.ylabel("Monthly Spend")
plt.title("K-Means Clustering")
plt.show()
```

```python
score = silhouette_score(X, df["Cluster"])
print("Silhouette Score =", score)
```
